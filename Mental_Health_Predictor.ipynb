```python
lable_col=['Stress_Level']
ohe_col=cat_cols.drop(lable_col).to_list()
ohe_col
```

```python
X=df.drop(columns=['Mental_Health_Score','Country'])
X
```

```python
y=df['Mental_Health_Score']
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
X_train
```

```python
skewed_col=['Study_Hours']

other_num_col=numeric_columns.drop(columns=skewed_col).drop(columns=['Mental_Health_Score']).columns.to_list()

lable_col

ohe_col

other_num_col

feature_cols=skewed_col+other_num_col+lable_col+ohe_col

feature_cols

other_num_col

len(ohe_col)

lable_col

ohe_col
```

```python
from sklearn.pipeline import Pipeline
```

```python
from sklearn.preprocessing import StandardScaler,OrdinalEncoder,OneHotEncoder,FunctionTransformer
```

```python
skew_pipe=Pipeline(steps=[
    ('log',FunctionTransformer(np.log1p)),
    ('scaler',StandardScaler())
])

other_num_pipe=Pipeline(steps=[
    ('scaler',StandardScaler())
])

lable_pipe=Pipeline(steps=[
    ('le',OrdinalEncoder(categories=[order]))
])

ohe_pipe=Pipeline(steps=[
    ('ohe',OneHotEncoder(handle_unknown='ignore'))
])
skew_pipe
```

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessor=ColumnTransformer(transformers=[
    ('skew',skew_pipe,skewed_col),
    ('other',other_num_pipe,other_num_col),
    ('le',lable_pipe,lable_col),
    ('ohe',ohe_pipe,ohe_col)
])

preprocessor
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
lr=Pipeline(steps=[
    ('preprocessing',preprocessor),
    ('regressor',LinearRegression())
])
```

```python
lr.fit(X_train,y_train)
y_pred1=lr.predict(X_test)
y_pred3=lr.predict(X_train)
lr_r2_testing=r2_score(y_test,y_pred1)
lr_r2_training=r2_score(y_train,y_pred3)
lr_mae=mean_squared_error(y_pred1,y_test)
print('r2:',r2_score(y_test,y_pred1))
print('mse:',mean_squared_error(y_pred1,y_test))
print('mae:',mean_absolute_error(y_pred1,y_test))
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf=Pipeline(steps=[
    ('preprocessing',preprocessor),
    ('random forest',RandomForestRegressor())
])
```

```python
rf.fit(X_train,y_train)
y_pred2=rf.predict(X_test)
y_pred4=rf.predict(X_train)
rf_r2_testing=r2_score(y_test,y_pred2)
rf_r2_training=r2_score(y_train,y_pred4)
rf_mae=mean_squared_error(y_pred2,y_test)
print('r2:',r2_score(y_test,y_pred2))
print('mse:',mean_squared_error(y_pred2,y_test))
print('mae:',mean_absolute_error(y_pred2,y_test))
```

# Hyper Parameter Tunning

```python
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    'random forest__n_estimators'       : [100,200,300],
    'random forest__max_depth'         : [5,10,15],
    'random forest__min_samples_split' : [2,5,10],
    'random forest__min_samples_leaf'  : [1,2,4]
}

random_search = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_grid,
    n_iter = 15,
    cv = 5,
    scoring='r2',
    random_state=42,
    n_jobs =-1 #processor free
)

random_search.fit(X_train, y_train)
```

```python
random_search.best_params_
```

```python
rf_best_pipeline = random_search.best_estimator_
rf_best_preds    = rf_best_pipeline.predict(X_test)

print(f"R2 Score for random forest after hyperparameter tuning {r2_score(y_test, rf_best_preds)}")
print(f"MAE Score for random forest after hyperparameter tuning {mean_absolute_error(y_test, rf_best_preds)}")
```

```python
from sklearn.metrics import mean_squared_error

# Calculate RMSE for Linear Regression
lr_rmse = np.sqrt(mean_squared_error(y_test, y_pred1))

# Calculate RMSE for default Random Forest
rf_rmse = np.sqrt(mean_squared_error(y_test, y_pred2))

# Calculate RMSE for tuned Random Forest
rf_tuned_preds          = random_search.best_estimator_.predict(X_test)
rf_tuned_rmse           = np.sqrt(mean_squared_error(y_test, rf_tuned_preds))
rf_tuned_training_preds = random_search.best_estimator_.predict(X_train)
r2_tuned_training       = r2_score(y_train, rf_tuned_training_preds)
rf_tuned_mae            = mean_absolute_error(y_test, rf_tuned_preds)
rf_tuned_r2             = r2_score(y_test, rf_tuned_preds)


# Create a DataFrame to consolidate results
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest (default)', 'Random Forest (tuned)'],
    'R2': [lr_r2_testing, rf_r2_testing, rf_tuned_r2],
    'Training R2': [lr_r2_training, rf_r2_training, r2_tuned_training],
    'MAE': [lr_mae, rf_mae, rf_tuned_mae],
    'RMSE': [lr_rmse, rf_rmse, rf_tuned_rmse]
})

print(results)
```
